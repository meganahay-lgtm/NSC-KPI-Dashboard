```python
aroflo_df = pd.read_excel("../data/aroflo_invoicing_report.xlsx")

target_month = "Dec"  # this won't exist in Jupyter yet - for now, just hardcode "Dec"

# 1. Which assets are SCHEDULED for service in the target month?
scheduled_serials = set(nsc_df[nsc_df[target_month].notna()]["Serial Number"])

# 2. Which assets were actually INVOICED for preventative service, in that month, this year?
aroflo_df["Invoice Date"] = pd.to_datetime(aroflo_df["Invoice Date"])
this_month_prevent = aroflo_df[
    (aroflo_df["Job Type"] == "Preventative Service") &
    (aroflo_df["Invoice Date"].dt.strftime("%b") == target_month) &
    (aroflo_df["Invoice Date"].dt.year == 2025)
]
invoiced_serials = set(this_month_prevent["Asset Serial Number"])

# 3. On time = scheduled AND invoiced
on_time = scheduled_serials & invoiced_serials
servicing_on_time_rate = len(on_time) / len(scheduled_serials) * 100

print(len(scheduled_serials), len(invoiced_serials), len(on_time))
servicing_on_time_rate
```
